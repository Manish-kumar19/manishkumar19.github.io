### import labraries.

```python
import tensorflow as tf
import numpy as np
```

### Generating data

```python
xs = np.array([ -2.0,-1.0,  0.0, 1.0, 2.0], dtype=float)
ys = np.array([   19,   8, 1  , -2.0, -1.0], dtype=float)
```

```python
model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(units=1)
    ])

```

```python
model.compile(optimizer="sgd",loss="mean_squared_error")
```

### Training the data with different epochs

```python
def trainModel (eph) :
  model = tf.keras.Sequential([
    # Define the input shape
    tf.keras.Input(shape=(1,)),
    # Add a Dense layer
    tf.keras.layers.Dense(units=1)
    ])
  model.compile(optimizer='sgd', loss='mean_squared_error')
  model.fit(xs, ys, epochs=eph)
  return model
```

```python
model10 = trainModel(10)
```

```python
# make the prediction.

print(f"model predicted: {model10.predict(np.array([3.0]))}")
```

```python
print(f"model predicted: {model10.predict(np.array([3.0]), verbose=0).item():.5f}")
```

```python
model50 = trainModel(50)
```

```python
print(f"model predicted: {model50.predict(np.array([3.0]), verbose=0).item():.5f}")
```

# Linear data

```python
 # y = 2x - 1

x1s = np.array([ -2,-1,0,1,2 ], dtype=float)
y1s = np.array([-5,-3,-1,1,3 ], dtype=float)
```

```python
def trainModelLinear (eph):
  modelL = tf.keras.Sequential([
      tf.keras.Input(shape=(1,)),
      tf.keras.layers.Dense(units=1)
      ])
  modelL.compile(optimizer="sgd", loss="mean_squared_error")
  modelL.fit(x1s, y1s, epochs=eph)
  return modelL
```

```python
modelL10 = trainModelLinear(10)
```

```python
print(f"model predicted: {modelL10.predict(np.array([3.0]), verbose=0).item():.5f}")
```

```python
modelL50 = trainModelLinear(50)
print(f"model predicted: {modelL50.predict(np.array([3.0]), verbose=0).item():.5f}")
```

```python
modelL200 = trainModelLinear(200)
print(f"model predicted: {modelL200.predict(np.array([6.0]), verbose=2).item():.5f}")
```
